```python

import pandas as pd

data = {
    "Customer_Name": ["Rahul", "Amit", "Priya", "Rahul", "Neha"],
    "Age": [25, 30, None, 25, 28],
    "City": ["Jaipur", "Delhi", "Jaipur", "Jaipur", None],
    "Purchase": [1200, 2500, 1800, 1200, 3000]
}

df = pd.DataFrame(data)

# Display the dataset
df
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Dataset shape:", df.shape)
```

```python
clean_df = df.copy()
```

```python
clean_df = clean_df.drop_duplicates()
```

```python
clean_df["Age"] = clean_df["Age"].fillna(
    clean_df["Age"].median()
)
```

```python
clean_df["City"] = clean_df["City"].fillna("Unknown")
```

```python
clean_df
```

```python
print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("\nRemaining missing values:")
print(clean_df.isnull().sum())
```

```python
clean_df.to_csv("cleaned_customers.csv", index=False)
```

```python
clean_df
```

```python

# Create a before-and-after cleaning report

report = pd.DataFrame({
    "Metric": [
        "Original rows",
        "Cleaned rows",
        "Duplicate rows removed",
        "Missing values before cleaning",
        "Missing values after cleaning"
    ],
    "Value": [
        len(df),
        len(clean_df),
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        clean_df.isnull().sum().sum()
    ]
})

print("DATA CLEANING REPORT")
display(report)
```

```python

# Save the data cleaning report
report.to_csv("data_cleaning_report.csv", index=False)

print("Report saved successfully!")

# Display the report
display(report)
```
